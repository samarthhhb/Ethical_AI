Model Training - Logistic Regression Model for Vulnerability Detection.

<br>

Logistic Regression refers to a supervised machine learning algorithm which is used for binary classification which can be 0 or 1, True or False and if a condition is True or Not. Here in this particular context we use this same algorithm to identify weather the user is at high risk or not for a potential exploit/ threat by an LLM for the user data to be mapped against them.

```python
retry_threshold = df_interaction_log["retry_count"].quantile(0.75)
satisfaction_threshold = 4
drift_threshold = 0.7
```

```python
df_interaction_log["high_risk"] = (
    (df_interaction_log["retry_count"] > retry_threshold) &
    (df_interaction_log["satisfaction_score"] >= satisfaction_threshold) &
    (df_interaction_log["drift_score"] > drift_threshold)
).astype(int)

df_interaction_log["high_risk"].value_counts()
```

```python
X = df_interaction_log[['input_length',
       'response_length', 'response_time_ms', 'token_usage',
       'satisfaction_score', 'retry_count', 'model_version', 'device_type',
       'network_latency', 'ambiguity', 'creativity', 'factual_density',
       'personalization', 'anomaly_flag', 'outlier_score', 'drift_score'
]]

y = df_interaction_log["high_risk"]
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X = scaler.fit_transform(X)
y = scaler.fit_transform(y.values.reshape(-1, 1))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
model = LogisticRegression(class_weight='balanced')

pipe = make_pipeline(scaler, model)
pipe.fit(X_train, y_train)
```

Model Score - 95.5%

```python
model_score = pipe.score(X_test, y_test)
print("The model score is: ", model_score*100)
```

```python
from sklearn.metrics import confusion_matrix, mean_absolute_error, classification_report
y_pred = pipe.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print(classification_report(y_test, y_pred))
print('\nThe Mean Absolute Error is: ', mean_absolute_error(y_test, y_pred))
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, cmap='Blues', fmt='g')
plt.xlabel('Predicted value of High Risk in AI Systems')
plt.ylabel('Actual value of High Risk in AI Systems')
plt.title('Confusion Matrix for Logistic Regression Model')
```
